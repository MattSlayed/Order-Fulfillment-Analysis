```python
# Import libraries for data analysis and visualization
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime, timedelta

# Set random seed for reproducibility
np.random.seed(42)
```

```python
# Create simulated order fulfillment data
n = 1000  # Number of orders
data = {
    'OrderID': range(1, n+1),
    'OrderDate': [datetime(2024, 1, 1) + timedelta(days=np.random.randint(0, 365)) for _ in range(n)],
    'ProcessingTime': np.random.normal(24, 5, n),  # Mean: 24 hours, SD: 5 hours
    'ShippingTime': np.random.normal(48, 10, n),   # Mean: 48 hours, SD: 10 hours
    'DeliveryTime': np.random.normal(72, 15, n),   # Mean: 72 hours, SD: 15 hours
    'CustomerSatisfaction': np.random.randint(1, 6, n)  # Ratings 1–5
}
df = pd.DataFrame(data)

# Save the dataset
df.to_csv('order_fulfillment.csv', index=False)
print("Dataset created and saved as 'order_fulfillment.csv'")
```

```python
# Display first few rows
print("First 5 rows of the dataset:")
print(df.head())

# Check for missing values
print("\nMissing values:")
print(df.isnull().sum())

# Summary statistics
print("\nSummary statistics:")
print(df.describe())
```

```python
# Calculate average times for each stage
avg_times = df[['ProcessingTime', 'ShippingTime', 'DeliveryTime']].mean()
print("\nAverage times (hours):")
print(avg_times)

# Identify bottlenecks (stages with average time > 48 hours)
bottlenecks = avg_times[avg_times > 48]
print("\nBottlenecks (stages > 48 hours):")
print(bottlenecks)
```

```python
# Plot distribution of processing time
plt.figure(figsize=(10, 6))
sns.histplot(df['ProcessingTime'], kde=True, color='blue')
plt.title('Distribution of Order Processing Time')
plt.xlabel('Processing Time (Hours)')
plt.ylabel('Frequency')
plt.savefig('processing_time_dist.png')
plt.close()
print("Saved: processing_time_dist.png")
```

```python
# Scatter plot of customer satisfaction vs. delivery time
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='DeliveryTime', y='CustomerSatisfaction', alpha=0.5)
plt.title('Customer Satisfaction vs. Delivery Time')
plt.xlabel('Delivery Time (Hours)')
plt.ylabel('Customer Satisfaction (1–5)')
plt.savefig('satisfaction_vs_delivery.png')
plt.close()
print("Saved: satisfaction_vs_delivery.png")
```

```python
# Save analysis summary
with open('process_summary.txt', 'w') as f:
    f.write("Order Fulfillment Process Analysis\n\n")
    f.write("Average Times (hours):\n")
    f.write(str(avg_times) + "\n\n")
    f.write("Bottlenecks (stages > 48 hours):\n")
    f.write(str(bottlenecks) + "\n\n")
    f.write("Recommendations:\n")
    f.write("- Automate order processing to reduce time to <20 hours.\n")
    f.write("- Partner with faster shipping providers to cut delivery time.\n")
    f.write("- Communicate expected delivery times to improve satisfaction.")
print("Saved: process_summary.txt")
```
